```python
import ColabTurtle.Turtle as t
from IPython.display import clear_output
```

```python
def create_l_system(iterations, axiom, rules):
    """Generate L-System instructions based on axiom and rules."""
    result = axiom
    for _ in range(iterations):
        new_string = ""
        for char in result:
            new_string += rules.get(char, char)
        result = new_string
    return result

def draw_l_system(instructions, angle, distance):
    """Draw the L-System using turtle graphics.

    Parameters:
    - instructions: string of L-System commands
    - angle: turning angle in degrees
    - distance: forward movement distance
    """
    stack = []
    depth = 0
    for cmd in instructions:

      if cmd == 'F':
        # Change color based on branch depth
        if depth == 0:
          t.pencolor("brown")
        elif depth == 1:
          t.pencolor("green")
        elif depth == 2:
          t.pencolor("blue")
        elif depth == 3:
          t.pencolor("yellow")
        elif depth == 4:
          t.pencolor("orange")

        t.forward(distance)

      elif cmd == 'f':  # Move forward and draw
        t.penup()
        t.forward(distance)
        t.pendown()

      elif cmd == '+':  # Turn right
          t.right(angle)
      elif cmd == '-':  # Turn left
          t.left(angle)
      elif cmd == '[':  # Save current state, and depth
          stack.append((t.position(), t.heading(), depth))
          depth +=1 # Enter a new branch
      elif cmd == ']':  # Restore previous state, and depth
          position, heading, depth = stack.pop()
          t.penup()
          t.goto(position)
          t.setheading(heading)
          t.pendown()

def setup_turtle():
    t.initializeTurtle()
    t.hideturtle()
    t.speed(13)  # Fastest speed
    t.penup()

    # Start near the bottom center
    t.goto(t.window_width() // 2, 50)

    # Point upward
    t.setheading(90)

    t.pendown()
```

```python
# Visual Pattern 1
# tree parameters
tree_axiom = "F"
tree_rules = {"F": "F[+F]F[-F]F"}
tree_iterations = 4
tree_angle = 25
tree_distance = 8

# Generate and draw
setup_turtle()
tree_instructions = create_l_system(tree_iterations, tree_axiom, tree_rules)
draw_l_system(tree_instructions, tree_angle, tree_distance)

```

```python
# Visual Pattern 2
# plant parameters
plant_axiom = "X"
plant_rules = {"X": "F+[[X]-X]-F[-FX]+X",
               "F": "FF"
               }
plant_iterations = 4
plant_angle = 25
plant_distance = 5

# Generate and draw
setup_turtle()
plant_instructions = create_l_system(plant_iterations, plant_axiom, plant_rules)
draw_l_system(plant_instructions, plant_angle, plant_distance)

```
